```python
# Check the key - if you're not using sarvamai, check whichever key you're using! Ollama doesn't need a key.

import os
sarvamai_api_key = os.getenv('SARVAM_API_KEY')

if sarvamai_api_key:
    print(f"sarvamai API Key exists and begins {sarvamai_api_key[:8]}")
else:
    print("sarvamai API Key not set - please head to the troubleshooting guide in the setup folder")
```

```python
# And now - the all important import statement
# If you get an import error - head over to troubleshooting in the Setup folder
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why
# importing sarvamai instead of openai
from sarvamai import SarvamAI
```

```python
client = SarvamAI(api_subscription_key=sarvamai_api_key)
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "What is 2+2?"}]
```

```python
response = client.chat.completions(messages=messages)
print(response.choices[0].message.content)
```

```python
# And now - let's ask for a question:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# ask the question to the model and print the response
response = client.chat.completions(messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# form a new message list with the new question, and ask the model to answer it
messages = [{"role": "user", "content": question}]
```

```python
# ask it again
response = client.chat.completions(messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown
Markdown(f"**Question:** {question}  \n**Answer:** {answer}")
```

```python
# pick a business area that might be worth exploring with an agentic ai opportunity
# ask llm to present a pain point in that industry - something challenging that might be ripe for an agentic solution
# Finally have a third LLM call propose the agentic AI solution to that problem.
```

```python
# ask the question to the model and print the response
messages = [{"role": "user", "content": "Pick a business area that might be worth exploring with an agentic AI opportunity, and present a pain point in that industry - something challenging that might be ripe for an agentic solution. Respond in the format 'Business Area: _____ Pain Point: _____' and respond only with that."}] 

response = client.chat.completions(messages=messages)

business_and_pain = response.choices[0].message.content
print(business_and_pain)
```

```python
messages = [{"role": "user", "content": f"Given the business area and pain point: {business_and_pain}, propose an agentic AI solution to that problem. Be creative and bold in your response!"}]
response = client.chat.completions(messages=messages)
solution = response.choices[0].message.content
print(solution)
```

```python
from IPython.display import Markdown
Markdown(f"**Business and Pain Point:** {business_and_pain}  \n**Proposed Agentic AI Solution:** {solution}") 
```
