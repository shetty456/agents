```python
# start with imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display
from sarvamai import SarvamAI
```

```python
load_dotenv(override=True)
```

```python
sarvamai_api_key = os.getenv("SARVAM_API_KEY")

if sarvamai_api_key:
    print("API key loaded successfully.")
else:
    print("Failed to load API key. Please check your .env file.")
```

```python
request = "Please come up with a challenging, nuanced question that I can ask a number of LLMs to evaluate their intelligence. "
request += "Answer only with the question, no explanation."
messages = [{"role": "user", "content": request}]
```

```python
messages
```

```python
client = SarvamAI(api_subscription_key=sarvamai_api_key)
```

```python

response = client.chat.completions(messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
display(Markdown(f"**Generated Question for Evaluating LLM Intelligence:**  \n{question}"))
```
